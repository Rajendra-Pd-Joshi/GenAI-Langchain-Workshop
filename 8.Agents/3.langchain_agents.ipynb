```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.tools import tool

@tool
def calculate_tipping(bill_amount: float, tip_percentage: float = 15.0) -> float:
    """Calculates the total tip amount based on a bill and percentage."""
    return bill_amount * (tip_percentage / 100)

tools = [calculate_tipping]
```

```python
from langchain.agents import create_agent

# Initialize your agent harness
agent = create_agent(
    model="openai:gpt-4o-mini",  # Choose your provider and model name
    tools=tools,
    system_prompt="You are a helpful restaurant assistant. Use tools when calculations are required."
)
```

```python
from langchain_core.messages import HumanMessage
```

```python
messages = [HumanMessage(content= "How much tip should I leave on a $85 bill if I want to give 18%?")]
```

```python
response = agent.invoke({
    'messages':messages
})
```
