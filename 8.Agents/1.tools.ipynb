# Tools and Agents

## What is a Tool?

A **tool** is simply a Python function (or API) that is packaged in a way the **LLM can understand and call when needed**.

LLMs like GPT are great at:

- Reasoning
- Language generation

But an LLM by itself can't reliably do things like:

- Access live data (weather, news)
- Do reliable math
- Call APIs
- Run code
- Interact with a database

Tools extend the capabilities of an LLM by allowing it to interact with the outside world.

---

## Tool + LLM = Tool Calling

The basic flow is:

```text
Tools → Tool Calling → Agents
  ↑         ↑
  │         │
  └─────────┘


The LLM decides when a tool is needed, selects the appropriate tool, provides the required inputs, and uses the tool's result.

What is an Agent?

An AI agent is an LLM-powered system that can autonomously think, decide, and take actions using external tools or APIs to achieve a goal.

An agent combines two major capabilities:

┌─────────────────────────────────────┐
│                Agent                │
│                                     │
│  ┌────────────────┐ ┌─────────────┐ │
│  │ Reasoning &    │ │   Action    │ │
│  │ Decision       │ │             │ │
│  │ Making         │ │             │ │
│  └────────────────┘ └─────────────┘ │
│                                     │
│       LLM              Tools        │
└─────────────────────────────────────┘

Components
LLM — Reasoning

The LLM is responsible for:

Understanding the user's goal
Reasoning about the problem
Deciding what action to take
Choosing which tool to use
Tools — Action

Tools allow the agent to actually perform actions, such as:

Searching the web
Calling an API
Running code
Sending an email
Querying a database
Built-in Tools

A built-in tool is a tool that LangChain already provides for you.

It is:

Pre-built
Production-ready
Requires minimal or no setup

You don't have to write the function logic yourself — you can simply import and use it.

Tools vs Agents

A useful way to think about the relationship is:

Tool
  ↓
A capability that can perform one specific task

LLM + Tool Calling
  ↓
Allows the LLM to decide when and how to use tools

Agent
  ↓
Combines reasoning + decision making + tool usage
to achieve a goal autonomously

Simple Example

Suppose the user asks:

"What's the weather in London today?"

The LLM knows the question requires live information, so it can:

Understand the request
Decide that a weather tool is required
Call the weather API/tool
Receive the result
Use the result to generate a response

The key idea is that the LLM provides the reasoning, while tools provide the ability to take actions and access external systems.

# 3. Wikipedia


````python
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper

wiki_wrapper = WikipediaAPIWrapper(
    top_k_results=2,
    doc_content_chars_max=4000
)

wiki_tool = WikipediaQueryRun(
    api_wrapper=wiki_wrapper
)

result = wiki_tool.invoke("Artificial Intelligence")

print(result)
````

# 4. Python REPL Tool
A Python REPL tool allows an agent to execute Python code.

For example, the LLM can decide to use Python when it needs to perform calculations.

````python
from langchain_experimental.tools import PythonREPLTool

python_tool = PythonREPLTool()

result = python_tool.invoke(
    "print(sum(range(1, 101)))"
)

print(result)
````

````python
from langchain_core.tools import tool
````

## 1. define a function

````python
def multiply(a,b):
    """ this function will multiply two integers and return an integr"""
    return a*b
````

## 2. add type hinting

````python
def multiply(a:int,b:int)->int:
    """this function will multiply two integers and return an integer"""
    return a*b
````

## 3. add the @tool decorator

````python
@tool
def multiply(a:int,b:int)->int:
    """this function will multiply two integers and return an integer"""
    return a*b
````

````python
multiply.invoke({'a':2,'b':3})
````

````python
multiply.args_schema.model_json_schema()
````

# Method 2 - Using StructuredTool

````python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel,Field
````

````python
class multiplySchema(BaseModel):
    a:int = Field(default=1,description='first number to multiply')
    b:int = Field(default=1,description='second number to multiply')
````

````python
def multiply_func(a:int,b:int)->int:
    return a*b
````

````python
multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    description='the tool is used to multiply two integers',
    args_schema=multiplySchema
)
````

````python
multiply_tool.invoke({'a':3,'b':5})
````

# Method 3 - Using BaseTool Class

````python
from langchain.tools import BaseTool
from typing import Type
````

````python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
````

````python
class MultiplyTool(BaseTool):
    name:str = 'multiply'
    description:str = 'multiply two numbers'
    args_schema : Type[BaseModel] = MultiplyInput
    def _run(self,a:int,b:int)->int:
        return a*b
````

````python
multiplytool = MultiplyTool()
````

````python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
````

# Toolkit

````python
from langchain_core.tools import tool
````

````python
@tool
def add(a:int,b:int)->int:
    """ this tool is used to add two integers and returns and integer"""
    return a+b
````

````python
@tool
def multiply(a:int,b:int)->int:
    """ this tool is used to multiply two integers and returns and integer"""
    return a*b
````

````python
class MathToolKit:
    def get_tools(self):
        return [add,multiply]
````

````python
toolkit = MathToolKit()
tools = toolkit.get_tools()
````

````python
for tool in tools:
    print(tool.name,"->",tool.description)
````
