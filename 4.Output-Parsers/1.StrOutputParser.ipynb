# StrOutputParser

The StrOutputParser is the simplest output parser in langchain. it is used to parse the output of a llm and return it as a plain string

why we need this when we can do response.content

topic -> llm -> detailed report -> llm2 -> summary 

if in this flow we need to just parse the output before passing to the llm2 do we can user stroutput parser in the chain

prompt -> llm ->parser -> output

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
```

```python
model = ChatOpenAI()
```

```python
parser = StrOutputParser()
```

```python
template1 = PromptTemplate(
    template = 'give a brief research on the following topic {topic}',
    input_variables=['topic']
)
```

```python
template2 = PromptTemplate(
    template='give a 5 line summary from teh following research {research}',
    input_variables=['research']
)
```

```python
chain = template1 | model | parser | template2 | model | parser
```

```python
result = chain.invoke({
    'topic':'black hole'
})
```
