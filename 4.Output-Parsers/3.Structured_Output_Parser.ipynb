## Structured Output Parser

we can enforce an schema to follow

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_classic.output_parsers import StructuredOutputParser,ResponseSchema
```

```python
parser = StructuredOutputParser(response_schemas=[ResponseSchema(name='fact-1',description='the first and most important fact'),
                                                  ResponseSchema(name='fact-2',description='the second fact')
                                                ])
```

```python
model = ChatOpenAI()
```

```python
prompt = PromptTemplate(
    template='give the information about the batting class of {batsman}\n{format_instruction}',
    input_variables=['batsman'],
    partial_variables={'format_instruction':parser.get_format_instructions}
)
```

```python
chain = prompt | model | parser
```

```python
chain.invoke(input={
    'batsman':'steve smith'
})
```

### we can enforce a llm to give response on this schema but we cannot validate the schema that is the major draw back of the structured output parsers
