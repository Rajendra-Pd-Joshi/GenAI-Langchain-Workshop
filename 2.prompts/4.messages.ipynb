```python
from langchain_core.messages import HumanMessage, AIMessage,SystemMessage
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
model = ChatOpenAI(model_name= "gpt-4o-mini", temperature=0.5)
```

```python
messages =[
    SystemMessage(content='you are a help ful asssistant'),
    HumanMessage(content='write a 5 line poem about steve smith batting class')
]
```

```python
messages.append(AIMessage(content=model.invoke(messages).content))
```

```python
print(messages)
```
