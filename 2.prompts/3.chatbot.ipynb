```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
model = ChatOpenAI()
```

```python
while True:
    user = input('you: ')
    if user.lower() == 'exit':
        break
    print('you: ',user)
    response = model.invoke(user).content
    print('bot: ',response)
```

## building another chatbot with some sort of memory


```python
chat_history = []
while True:
    query = input('Enter your query: ')
    if query.lower() == 'exit':
        break
    print('you: ', query)
    chat_history.append(query)
    response = model.invoke(chat_history).content
    chat_history.append(response)
    print('bot: ', response)

print(chat_history)
```
