# TypedDict vs Pydantic vs JSON Schema

## Use `TypedDict` if:

* You only need **type hints** and basic structure enforcement.
* You don't need **data validation** (for example, checking that numbers are positive).
* You trust the LLM to return correct data.

---

## Use `Pydantic` if:

* You need **data validation** (for example, sentiment must be `"positive"`, `"neutral"`, or `"negative"`).
* You need **default values** if the LLM misses fields.
* You want **automatic type conversion** (for example, `"100"` → `100`).

---

## Use `JSON Schema` if:

* You don't want to import extra Python libraries such as Pydantic.
* You need validation but don't need Python objects.
* You want to define a structure using a **standard JSON format**.

---

## Quick Comparison

| Feature              | TypedDict | Pydantic | JSON Schema               |
| -------------------- | --------- | -------- | ------------------------- |
| Type hints           | ✅         | ✅        | ✅                         |
| Data validation      | ❌         | ✅        | ✅                         |
| Default values       | Limited   | ✅        | ✅                         |
| Type conversion      | ❌         | ✅        | Depends on implementation |
| Python objects       | ✅         | ✅        | ❌                         |
| Standard JSON format | ❌         | ❌        | ✅                         |

### Simple Rule

```text
Need basic structure?
        ↓
    TypedDict

Need validation + type conversion?
        ↓
     Pydantic

Need a standard JSON-based schema?
        ↓
    JSON Schema
```
