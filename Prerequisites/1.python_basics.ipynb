# 3. Variables in Python

A variable is a name that refers to a value.

Python does not require us to explicitly declare the data type.

```python
name = "Rajendra"
age = 23
height = 5.8
```

We can print the variables:

```python
print(name)
print(age)
print(height)
```

The type is determined automatically.

```python
x = 10
print(type(x))

x = "Hello"
print(type(x))
```

Output:

```text
<class 'int'>
<class 'str'>
```

The same variable can refer to values of different types at different times.

---

# 4. Rules for Naming Variables

Variable names:

* Must start with a letter or `_`
* Cannot start with a number
* Can contain letters, numbers and `_`
* Are case-sensitive
* Cannot use Python keywords

Valid:

```python
name = "Rajendra"
student_name = "Rajendra"
age2 = 23
_private_value = 100
```

Invalid:

```python
2age = 23
student-name = "Rajendra"
class = "Python"
```

Python variable names are case-sensitive:

```python
name = "Rajendra"
Name = "John"

print(name)
print(Name)
```

These are two different variables.

# 10. Arithmetic Operators

Arithmetic operators perform mathematical operations.

| Operator | Meaning        |
| -------- | -------------- |
| `+`      | Addition       |
| `-`      | Subtraction    |
| `*`      | Multiplication |
| `/`      | Division       |

# 11. Comparison Operators

Comparison operators compare values.

They return either:

```text
True
```

or:

```text
False
```

Operators:

```text
==    Equal
!=    Not equal
>     Greater than
<     Less than
>=    Greater than or equal
<=    Less than or equal
```

Example:

```python
a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

Output:

```text
False
True
False
True
False
True
```

These operators are especially important when writing conditions.

---

# 12. Logical Operators

Logical operators combine conditions.

| Operator | Meaning                             |
| -------- | ----------------------------------- |
| `and`    | Both conditions must be True        |
| `or`     | At least one condition must be True |
| `not`    | Reverses the result                 |

Example:

```python
age = 23
has_id = True

print(age >= 18 and has_id)
```

Output:

```text
True
```

Another example:

```python
is_student = True
is_employee = False

print(is_student or is_employee)
print(not is_student)
```

Output:

```text
True
False
```

# 15. If-Else Statements

Conditional statements allow our program to make decisions.

Basic structure:

```python
if condition:
    # code
else:
    # code
```

Example:

```python
age = 20

if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

Output:

```text
You are an adult.
```

### Important: Indentation

Python uses indentation to define blocks.

Correct:

```python
if age >= 18:
    print("Adult")
```

Incorrect:

```python
if age >= 18:
print("Adult")
```

Indentation is part of Python syntax.

# 16. `if`, `elif`, and `else`

We can check multiple conditions.

```python
marks = 85

if marks >= 90:
    print("A+")
elif marks >= 80:
    print("A")
elif marks >= 70:
    print("B")
else:
    print("C")
```

Output:

```text
A
```

Python checks conditions from top to bottom.

Once a condition is `True`, its block is executed and the remaining conditions are skipped.

# 22. `range()`

`range()` generates a sequence of numbers.

```python
for i in range(5):
    print(i)
```

# 23. While Loop

A `while` loop continues while a condition is `True`.

```python
count = 0

while count < 3:
    print("Hello")
    count += 1
```

Output:

```text
Hello
Hello
Hello
```

Be careful with `while` loops.

If the condition never becomes `False`, the loop can run indefinitely.

# 24. Break and Continue

## `break`

Stops the loop.

```python
for i in range(10):

    if i == 5:
        break

    print(i)
```

Output:

```text
0
1
2
3
4
```

## `continue`

Skips the current iteration.

```python
for i in range(5):

    if i == 2:
        continue

    print(i)
```

Output:

```text
0
1
3
4
```

# 27. Return Values

A function can return a value.

```python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```

Output:

```text
30
```

The difference between `print()` and `return` is important.

```python
def add(a, b):
    print(a + b)
```

prints the result.

Whereas:

```python
def add(a, b):
    return a + b
```

returns the result so that another part of the program can use it.

# 31. Exception Handling

Errors can occur while a program is running.

Python provides `try` and `except` for handling errors.

```python
try:
    number = int(input("Enter a number: "))
    print(10 / number)

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

This prevents the entire program from crashing for expected errors.

# 32. Working with JSON

JSON is extremely common in APIs and AI applications.

Example JSON:

```json
{
    "name": "Rajendra",
    "age": 23,
    "skills": ["Python", "LangChain"]
}
```

Python can work with JSON using the built-in `json` module.

```python
import json

data = {
    "name": "Rajendra",
    "age": 23
}

json_data = json.dumps(data)

print(json_data)
```

Convert JSON back to a Python object:

```python
data = json.loads(json_data)

print(data["name"])
```

JSON and Python dictionaries are closely related, which makes dictionaries especially important for API development.
